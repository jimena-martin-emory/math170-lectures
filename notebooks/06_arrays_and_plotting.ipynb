# 6.1 NumPy and Array Computing

To plot a curve $y = f(x)$, a computer draws straight lines between many points
on the curve, so first we must store the coordinates of those points: one
sequence of $x$-values and one sequence of the matching $y$-values, of equal
length. Lists can do this, but the **NumPy** package offers a data structure
made for the job, the **array**. (The plotting itself is done with `matplotlib`,
the standard plotting toolbox in Python; that is Section 6.2.)

NumPy is imported as

```python
import numpy as np
```

and every NumPy function is then called as `np.something`.

### Vectorize the bell curve

The bell curve (Gaussian) with center $m$ and width $s$ is

$$f(x) = \frac{1}{\sqrt{2\pi}\, s} \exp\left[-\frac{1}{2}\left(\frac{x-m}{s}\right)^2\right].$$

The function below only works for a single number. Write `gauss_v(x, m=0, s=1)`,
a vectorized version, and print its values at 9 equally spaced points in
$[-4, 4]$, with no `for` loop.

````python
import math

def gauss(x, m=0, s=1):
    return 1/(math.sqrt(2*math.pi)*s) * math.exp(-0.5*((x - m)/s)**2)

print(gauss(0))
````

# 6.4 Making a Movie of a Plot

A movie is a long sequence of pictures shown quickly one after the other. Here
the pictures are plots of the bell curve from Section 6.1,

$$f(x; m, s) = \frac{1}{\sqrt{2\pi}\, s} \exp\left[-\frac{1}{2}\left(\frac{x-m}{s}\right)^2\right],$$

as the width $s$ shrinks from 2 to 0.2. Run the setup cell first.

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x, m, s):
    return (1.0/(np.sqrt(2*np.pi)*s))*np.exp(-0.5*((x-m)/s)**2)

m = 0;  s_start = 2;  s_stop = 0.2
s_values = np.linspace(s_start, s_stop, 30)
x = np.linspace(m - 3*s_start, m + 3*s_start, 1000)
max_f = f(m, m, s_stop)    # f is largest at x = m, and larger for smaller s
````
