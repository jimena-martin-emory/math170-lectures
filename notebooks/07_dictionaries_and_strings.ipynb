# 7.1 Dictionaries

A list maps the integers `0, 1, 2, ...` to its elements. A **dictionary** is a
generalization of a list: it maps **keys** to values, and the keys do not have to
be integers. Most often they are strings:

```python
d = {'Norway': 'Oslo', 'Sweden': 'Stockholm', 'France': 'Paris'}
```

Each `key: value` pair is separated from the next by a comma, and the whole
dictionary sits inside curly braces `{ }`.

````python
# predict: ...
t = '1st line\n2nd line\n3rd line'
print(t)
print(t.split('\n'))
t_win = '1st line\r\n2nd line\r\n3rd line'    # Windows line endings
print(t_win.split('\n'))
print(t_win.splitlines())

# YOUR CODE HERE -- note what you changed
````

> **Note: where does the `'\r'` come from?**
>
> Operating systems mark the end of a line differently. macOS and Linux use one
> character, `'\n'` (newline). Windows uses two, `'\r\n'`: a *carriage return*
> `'\r'` followed by `'\n'`, a habit inherited from typewriters, which first
> returned the carriage to the left margin and then fed the paper up one line.
>
> `split('\n')` cuts only at the `'\n'`, so the `'\r'` just before each cut stays
> at the end of the piece on its left: `'1st line\r' | '2nd line\r' | '3rd line'`.
> `splitlines()` recognizes `'\n'`, `'\r\n'` and `'\r'` as line endings, so it works
> wherever the text came from.
>
> The leftover `'\r'` is invisible when printed, but it is still there:
> `'1st line\r' == '1st line'` is `False`, and a dictionary key with a hidden
> `'\r'` will not match when you look it up.
>
> When you read a file with `open(..., 'r')`, Python already turns every `'\r\n'`
> into `'\n'`, so you will rarely meet this there. It shows up in text that
> arrives some other way: pasted in, downloaded from the web, or read in binary
> mode.
